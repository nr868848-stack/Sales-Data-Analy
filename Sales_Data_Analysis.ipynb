```python
import pandas as pd
import numpy as np

np.random.seed(42)

n = 1000

data = {
    "Order_ID": range(5001, 5001 + n),
    "Customer_ID": np.random.randint(1001, 1201, n),
    "Order_Date": np.random.choice(
        pd.date_range("2025-01-01", "2025-12-31"),
        n
    ),
    "Product": np.random.choice(
        ["Laptop", "Mobile", "Headphones",
         "Keyboard", "Mouse", "Monitor"], n
    ),
    "Category": np.random.choice(
        ["Electronics", "Accessories"], n
    ),
    "Region": np.random.choice(
        ["North", "South", "East", "West"], n
    ),
    "Quantity": np.random.randint(1, 6, n),
    "Unit_Price": np.random.choice(
        [500, 1000, 1500, 3000, 5000,
         10000, 20000, 40000], n
    ),
    "Discount": np.random.choice(
        [0, 5, 10, 15, 20], n
    ),
    "Payment_Method": np.random.choice(
        ["UPI", "Credit Card", "Debit Card", "Cash"], n
    ),
    "Delivery_Days": np.random.randint(1, 11, n),
    "Rating": np.random.choice([1, 2, 3, 4, 5], n)
}

df = pd.DataFrame(data)

df["Sales"] = (
    df["Quantity"] *
    df["Unit_Price"] *
    (1 - df["Discount"] / 100)
)

df["Profit"] = df["Sales"] * np.random.uniform(0.05, 0.25, n)

df = df.sort_values("Order_Date").reset_index(drop=True)

df.head()
```
